# Survey Response Auto-Fill with Sentiment Analysis

This notebook trains a simple multi-output classifier that maps free-text feedback to structured survey answers using VADER sentiment scores + TF-IDF features.

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if (ROOT / "survey_sentiment").exists():
    sys.path.insert(0, str(ROOT))
elif (ROOT.parent / "survey_sentiment").exists():
    sys.path.insert(0, str(ROOT.parent))

from survey_sentiment.src.train import train, save_bundle
from survey_sentiment.src.survey_filler import SurveyFiller
```

```python
DATA = Path("survey_sentiment/data/sample_feedback.csv")
MODEL = Path("survey_sentiment/models/survey_filler.joblib")

bundle = train(DATA, classifier="logistic")
save_bundle(bundle, MODEL)
bundle["metrics"]
```

```python
filler = SurveyFiller(MODEL)

examples = [
    "Absolutely fantastic service and a beautiful product.",
    "Terrible quality. Support never replied.",
    "Arrived on time and works as described. Average overall.",
]

filler.fill_many(examples)
```

```python
# Rule-based baseline (no ML): map VADER compound → Likert answers
filler.rule_based_fill("Pretty good overall, would buy again.")
```
