```python
# function to convert the set of text into lower case 
def pre_process(data):
    df_pm = data.lower()
    return df_pm
    
```

```python
import string as str
def remove_specichar(data):
    symbols = {'~', ':', "'", '+', '[', '\\', '@', '^', '{', '%', '(', '-', '"', '*', '|', ',', '&', '<', '`', '}', '.', '_', '=', ']', '!', '>', ';', '?', '#', '$', ')', '/'}
    for i in data:
        print(i)
        if i in symbols:            
            df_pm = data.replace(i,'')
       
            return df_pm
```

```python
import re
def sep_rem(data):
    df_pm = re.sub(r"[^a-zA-Z0-9]"," ",data)
    df_pm1 = re.sub(r"[-()\"#/@;:<>{}/''`+=~|.!?,]", " ", df_pm)
    return df_pm1
```

```python
df_pm['clean'] = df_pm['text'].apply(lambda x:pre_process(x))
```

```python
df_pm['clean'][0]
```

```python
df_pm['cleansp']= df_pm['clean'].apply(lambda x:sep_rem(x))
```

```python
df_pm['cleansp'][0]
```

```python
tokenizer = AutoTokenizer.from_pretrained('nlptown/bert-base-multilingual-uncased-sentiment')
model = AutoModelForSequenceClassification.from_pretrained('nlptown/bert-base-multilingual-uncased-sentiment')
```

```python
def sentimnet_analysis(clean):
    token = tokenizer.encode(clean,return_tensors='pt')
    results = model(token)
    return int(torch.argmax(results.logits))
    
```

```python
sentimnet_analysis(df_pm['cleansp'].iloc[444])
```
